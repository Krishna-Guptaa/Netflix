## **Top 10 Netflix Genres**

```python
genres = df["listed_in"].str.split(", ").explode()

top_genres = genres.value_counts().head(10)

print(top_genres)
```

```python
sns.barplot(x = top_genres.values, y= top_genres.index, palette = "gist_heat")

plt.title("Top 10 Netflix Genres")
plt.xlabel("Number of Titles")
plt.ylabel("Genres")

plt.show
```

## **Top 10 Countries by Netflix Content**

```python
top_countries = df["country"].value_counts().head(5)
print(top_countries)
```

```python
sns.barplot(x = top_countries.values, y = top_countries.index, palette = "inferno")

plt.title("Top 10 Countries by Netflix Content")
plt.xlabel("Number of titles")
plt.ylabel("Country")

plt.show()
```

## **Distribution by Rating**

```python
order = df["rating"].value_counts()

print(order)
```

```python
plt.figure(figsize = (7,7))

sns.countplot(y = "rating",order = df["rating"].value_counts().index, data = df, palette = "rocket")

plt.title("Netflix Rating Distribution")
plt.xlabel("Number of Titles")
plt.ylabel("Rating")
plt.show()
```

## **Top 10 Directors**

```python
top_directors = df["director"].value_counts().head(10)

print(top_directors)
```

```python
sns.barplot(x = top_directors.values, y = top_directors.index, palette = "cubehelix")

plt.title("Top 10 Directors on Netflix")
plt.xlabel("Number of Titles")
plt.ylabel("Directors")
plt.show()
```
